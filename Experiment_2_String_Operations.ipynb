Write a Python program to accept a student's name, email address, and course name. Display the capitalized name, check the email domain, display the course length, and check whether the name contains only alphabets.

```python
name = input("Enter student name: ")
email = input("Enter email address: ")
course = input("Enter course name: ")

print("\n--- Student Information ---")

print("Capitalized Name:", name.capitalize())

if email.endswith("gmail.com"):
    print("Email Type: Gmail")
else:
    print("Email Type: Other")

print("Course Length:", len(course))

if name.isalpha():
    print("Name Validation: Valid")
else:
    print("Name Validation: Invalid")
```
