```python
attendance = 80
assignment_completed = True

if attendance >= 75:
    if assignment_completed:
        print("Student is eligible for the exam.")
    else:
        print("Complete the assignment first.")
else:
    print("Attendance is insufficient.")
```

```python
balance = 5000
withdrawal = 2000

if withdrawal > 0:
    if withdrawal <= balance:
        print("Withdrawal successful.")
    else:
        print("Insufficient balance.")
else:
    print("Invalid withdrawal amount.")
```

```python
marks = int(input("Enter marks: "))

if marks >= 90:
    print("Grade A+")
elif marks >= 80:
    print("Grade A")
elif marks >= 70:
    print("Grade B")
elif marks >= 60:
    print("Grade C")
elif marks >= 40:
    print("Grade D")
else:
    print("Fail")
```

Write a Python program to check whether a customer can withdraw a requested amount from their account.

```python
balance = float(input("Enter account balance: "))
amount = float(input("Enter withdrawal amount: "))

if amount > 0:
    if amount <= balance:
        print("Withdrawal successful")
        print("Remaining balance:", balance - amount)
    else:
        print("Insufficient balance")
else:
    print("Invalid amount")
```
