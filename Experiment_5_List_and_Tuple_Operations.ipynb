Create a list of marks. Find its length and arrange the marks in ascending and descending order.

```python
marks = [78, 92, 65, 88, 74]

print("Number of marks:", len(marks))

marks.sort()
print("Ascending order:", marks)

marks.sort(reverse=True)
print("Descending order:", marks)
```

Create a list from 1 to 20 and use list comprehension to display only the even numbers.

```python
numbers = list(range(1, 21))

even_numbers = [x for x in numbers if x % 2 == 0]

print("Even numbers:", even_numbers)
```
