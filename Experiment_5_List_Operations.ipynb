Create a list of marks. Find its length and arrange the marks in ascending and descending order.

```python
marks = [78, 92, 65, 88, 74]

print("Number of marks:", len(marks))

marks.sort()
print("Ascending order:", marks)

marks.sort(reverse=True)
print("Descending order:", marks)
```

Create a list of numbers. Count the occurrence of a particular number, find its index, and reverse the list.

```python
numbers = [5, 10, 15, 10, 20, 10, 25]

print("Count of 10:", numbers.count(10))
print("Index of 10:", numbers.index(10))

numbers.reverse()

print("Reversed list:", numbers)
```

Create a list from 1 to 20 and use list comprehension to display only the even numbers.

```python
numbers = list(range(1, 21))

even_numbers = [x for x in numbers if x % 2 == 0]

print("Even numbers:", even_numbers)
```

Create a list from 1 to 10 and create another list containing the squares of the numbers.

```python
numbers = list(range(1, 11))

squares = [x * x for x in numbers]

print("Numbers:", numbers)
print("Squares:", squares)
```
